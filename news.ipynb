```python
import feedparser
from transformers import pipeline
```

```python
feeds = {
    "Moneycontrol": "https://www.moneycontrol.com/rss/business.xml",
    "ET_Markets": "https://economictimes.indiatimes.com/markets/rssfeeds/1977021501.cms",
    "NSE_Corporate_Announcements": "https://www.nseindia.com/rss/corporate-announcements.xml",
    "NSE_Press_Releases": "https://www.nseindia.com/rss/nse-news.xml"
}
```

```python
sentiment_model = pipeline("sentiment-analysis", model = "ProsusAI/finbert")
```

```python
import feedparser

HEADERS = {
    "User-Agent": "Mozilla/5.0",
}

def get_news_sentiment():
    for name, url in feeds.items():
        print(f"--- Fetching from {name} ---")
        
        feed = feedparser.parse(url, request_headers=HEADERS)
        
        for entry in feed.entries[:5]:
            headline = entry.title.strip()
            sentiment = sentiment_model(headline)[0]
            
            print(f"News: {headline}")
            print(f"Sentiment: {sentiment['label']} ({sentiment['score']:.2f})")
            print(f"Link: {entry.link}\n")
```

```python
get_news_sentiment()
```

```python
import feedparser
from datetime import datetime

def get_stock_news_24h(ticker):
    url = f"https://news.google.com/rss/search?q={ticker}+stock+when:1d&hl=en-IN&gl=IN&ceid=IN:en"
    feed = feedparser.parse(url)
    
    print(f"--- Latest News for {ticker} (Past 24h) ---")
    for entry in feed.entries:
        print(f"Published: {entry.published}")
        print(f"Headline: {entry.title}")
        print(f"Link: {entry.link}\n")

get_stock_news_24h("RELIANCE")
```
